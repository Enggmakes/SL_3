```python
from itertools import combinations
```

```python
def load_itemset():
    return[['milk','bread','butter'],
           ['bread','milk'],
           ['butter','bread'],
           ['bread','milk','butter']]

def get_freq_itemset(dataset,min_support):
    item_counts={}
    for transactions in dataset:
        for item in transactions:
            item_counts[item]=item_counts.get(item,0)+1
    num_transactions=len(dataset)
    frequent_itemsets={
         (item,):count/num_transactions
          for item,count in item_counts.items()
          if count/num_transactions>=min_support
    }
    return frequent_itemsets
    
def has_infrequent_subset(candidate,prev_freq_itemsets):
    k=len(candidate)
    subsets=list(combinations(candidate,k-1))
    return any(tuple(sorted(sub)) not in prev_freq_itemsets for sub in subsets)


def generate_candidate(prev_freq_itemsets,k):
    candidates=[]
    items=list(prev_freq_itemsets.keys())
    for i in range(len(items)):
        for j in range(i+1,len(items)):
            l1=list(items[i])
            l2=list(items[j])
            l1.sort()
            l2.sort()
            candidate=tuple(sorted(set(l1) |set(l2)))
            if not has_infrequent_subset(candidate,prev_freq_itemsets):
                candidates.append(candidate)
    return candidates

def apriori(dataset,min_support):
    dataset=list(map(set,dataset))
    freq_itemsets={}
    one_itemset=get_freq_itemset(dataset,min_support)
    freq_itemsets.update(one_itemset)
    k=2
    current_freq_itemsets=one_itemset

    while current_freq_itemsets:
       candidates=generate_candidate(current_freq_itemsets,k)
       candidate_count={c:0 for c in candidates}

       for transactions in dataset:
           for candidate in candidates:
               if set(candidate).issubset(transactions):
                   candidate_count[candidate]+=1

       num_transactions=len(dataset)
       current_freq_itemsets={
           itemset:count/num_transactions
           for itemset,count in candidate_count.items()
           if  count/num_transactions>=min_support
       }
       freq_itemsets.update(current_freq_itemsets)
       k+=1
       return freq_itemsets
if __name__=="__main__":
    dataset=load_itemset()
    min_support=0.6
    freq_itemset=apriori(dataset,min_support)

    for itemset, support in freq_itemset.items():
        print(f"{itemset},{support}")
```
